## 1. Using Type Hints

Use type hints extensively to improve readability and maintainability.
Type hints should be as specific as possible.

### 1.1. Importing Annotations from Future

Although DeepTrack2 supports Python 3.9 and above, we consistently import
`annotations` from `__future__` to enable modern type hinting syntax that
is standard in Python 3.10 and later.

```python
# Must be the first import in the file.

from __future__ import annotations
```

### 1.2. Using Specific Type Hints

Suppose a function only accepts `"sum"` or `"average"`as valid values for a parameter.
The correct form of the type hint is ...

```python
# Correct

from typing import Literal


def process_values(mode: Literal["sum", "average"]) -> None:
    pass
```

... instead of using the general `str` type hint.

```python
# Incorrect


def process_values(mode: str) -> None:
    pass
```

This syntax restricts the parameter to only the specified values and improves type safety.

### 1.3. Specifying List Types

For example, when a function takes a list of integers, the type hint should be `list[int]` ...

```python
# Correct


def calculate_average(numbers: list[int]) -> None:
    pass
```

... instead of the general `list`, which does not specify the element type.

```python
# Incorrect


def calculate_average(numbers: list) -> None:
    pass
```

Prefer built-in generics (`list[int]`) over `typing.List[int]`.

### 1.4. Defining Class Attributes

Classes should define their attribute types before the `.__init__()` method to improve clarity and consistency.

```python
# Correct


class DataProcessor:

    mode: str
    values: list[float]

    def __init__(self: DataProcessor, mode: str, values: list[float]) -> None:
        self.mode = mode
        self.values = values
```

```python
# Incorrect


class DataProcessor:

    def __init__(self: DataProcessor, mode: str, values: list[float]) -> None:
        self.mode = mode
        self.values = values
```

## 2. Formatting Imports

Use absolute imports and avoid wildcard (`*`) imports.

Prefer absolute imports, e.g.:

    from deeptrack.features import Feature

Avoid:

    from ..features import Feature
    from deeptrack.features import *

**NOTE:** Prefer top-level imports. Import inside functions/classes only for
optional dependencies, expensive imports, or to avoid import cycles.

**NOTE:** Historically, we have accepted the use of wildcard (`*`) imports from directories within an `__init__.py` file. However, we are now phasing out this practice.

**NOTE:** Historically, we have accepted relative imports when the target is at the same level in the hierarchy inside a `__init__.py`. However, we are now phasing out this practice.

## 3. Writing Documentation

DeepTrack2 documentation should generally follow the [NumpyDoc style guide](https://numpydoc.readthedocs.io/en/latest/format.html#style-guide).

All modules, classes, methods, and functions should be documented.
The documentation should include a description of the class or method,
the parameters, the return value, and any exceptions that can be raised.

We appreciate efforts to improve documentation,
especially by adding examples that demonstrate typical usage.

**NOTE:** The NumpyDoc guide specifies using phrases such as `int or float`
rather than shorthand notations such as `int | float`.
Historically, we used to prefer the phrases in the docstrings.
However, we are now moving to the shorthand notations.

**NOTE:** In the type hints, we prefer `Literal["numpy", "torch"]` over `"numpy" | "torch"` because it is more compatible with IDEs and static type checkers. However, in the documentation, we prefer the shorthand form for compactness.

### 3.1. Writing Documentation for Functions and Methods

```python
from __future__ import annotations

from typing import Literal


def my_function(
    param1: int | float,
    param2: str,
    param3: Literal["numpy", "torch"] = "numpy",
    param4: list[int] | None = None,
) -> list[int]:
    """This is a short description of the function on one line.

    This is a longer description of the function. It should explain what the
    function does and how it works. Note also that each line is no longer than
    79 characters.

    Parameters
    ----------
    param1: int | float
        This is a description of the first parameter. It can be on multiple
        lines.
    param2: str
        This is a description of the second parameter.
    param3: "numpy" | "torch", optional
        This is a description of the third parameter. When the parameter is
        optional, its default value should be provided; in this case, "numpy".
    param4: list[int] | None, optional
        This is a description of the fourth parameter. Defaults to `None`.

    Returns
    -------
    list[int]
        This is a description of the return value. This can also be on multiple
        lines.

    Raises
    ------
    ValueError
        This is a description of the exception that can be raised. It can be
        on multiple lines.

    Examples
    --------
    >>> import deeptrack as dt

    Import DeepTRack2 as shown in the line above, if needed.

    This line of code uses the function:

    >>> my_function(1, "a")
    [1, 2, 3]

    """

    return [1, 2, 3]
```

### 3.2. Writing Documentation for Classes

```python
from __future__ import annotations


class MyClass:
    """A one-line description of the class.

    This is a longer description of the class and its methods. It should
    explain what the class does and how it works. Note also that each line is
    no longer than 79 characters.

    This extended description can also be in several paragraphs.

    Parameters
    ----------
    parameter_1: int
        This is a description of the first parameter. It can be on multiple
        lines.
    parameter_2: str, optional
        This is a description of the second parameter. When the parameter is
        optional, its default value should be provided, in this case "default".

    Attributes
    ----------
    attribute_1: int
        This is a description of the first attribute. It can be on multiple
        lines.
    attribute_2: str
        This is a description of the second attribute.

    Methods
    -------
    **Getter methods.**

    `get_1() -> int`
        Description of the method including the necessary details. It can be
        on multiple lines. Importantly the signature of the method should be
        on a single line (not exceeding 79 characters).
    `get_2() -> str`
        Description of the method.

    **Setter methods.**

    `set_1(new_value) -> None`
        Set first attribute.
    `set_2(new_value) -> None`
        Set second attribute.

    Examples
    --------
    Provide some simple examples of the use of this class.

    """

    # All attributes should be described here together with their types.
    attribute_1: int
    attribute_2: str

    def __init__(
        self: MyClass,
        parameter_1: int,
        parameter_2: str = "default",
    ) -> None:
        """Initialize class.

        ...

        """

        self.attribute_1 = parameter_1
        self.attribute_2 = parameter_2

    def get_1(
        self: MyClass,
    ) -> int:
        """..."""

        return self.attribute_1

    def get_2(
        self: MyClass,
    ) -> str:
        """..."""

        return self.attribute_2

    def set_1(
        self: MyClass,
        new_value: int,
    ) -> None:
        """..."""

        self.attribute_1 = new_value

    def set_2(
        self: MyClass,
        new_value: str,
    ) -> None:
        """..."""

        self.attribute_2 = new_value
```

## 4. Using Unit Tests (DeepTrack2 Conventions)

DeepTrack2 uses `unittest`.

Use one test file per module and one main test class per file.
Use one test method per class or function.
Disable docstring linting in tests using the standard `pylint` directives.
Prefer deterministic tests (set seeds explicitly).
If code supports torch, test both NumPy and torch paths gated by `TORCH_AVAILABLE`.

```python
# pylint: disable=C0115:missing-class-docstring
# pylint: disable=C0116:missing-function-docstring
# pylint: disable=C0103:invalid-name

from __future__ import annotations

import unittest

import numpy as np

from deeptrack import TORCH_AVAILABLE


if TORCH_AVAILABLE:
    import torch


class TestExample(unittest.TestCase):

    def test_example(self):
        rng = np.random.default_rng(0)
        x = rng.random((2, 3))
        self.assertEqual(x.shape, (2, 3))

        if TORCH_AVAILABLE:
            torch.manual_seed(0)
            x = torch.rand(2, 3)
            self.assertEqual(tuple(x.shape), (2, 3))
```

## 1. Using Type Hints

Use type hints extensively to make the code more readable and maintainable. The type hints should be as specific as possible.

### 1.1. Importing Annotations from Future

Although we support Python 3.9 and above, we regularly import `annotations` from `__future__` to enable modern type hinting syntax that is standard in Python 3.10 and later. To do so, add the following as the first import:

```python
from __future__ import annotations
```

### 1.2. Using Specific Type Hints

Suppose a function only accepts `"sum"` or `"average"` as valid values for a parameter. The correct form of the type hint is ...

```python
# Correct

from typing import Literal

def process_values(mode: Literal["sum", "average"]) -> float:
    pass
```

... instead of using the general `str` type hint.

```python
# Incorrect

def process_values(mode: str) -> float:
    pass
```

This syntax restricts the parameter to only the specified values and improves type safety.

### 1.3. Specifying List Types

For example, when a function takes a list of integers, the type hint should be `list[int]` ...

```python
# Correct

def calculate_average(numbers: list[int]) -> float:
    pass
```

... instead of the general `list`, which does not specify the element type.

```python
# Incorrect

def calculate_average(numbers: list) -> float:
    pass
```

### 1.4. Defining Class Attributes

Classes should define their attribute types before the `.__init__()` method to improve clarity and consistency.

```python
# Correct

class DataProcessor:

    mode: str
    values: list[float]

    def __init__(self: DataProcessor, mode: str, values: list[float]) -> None:
        self.mode = mode
        self.values = values
```

```python
# Incorrect

class DataProcessor:

    def __init__(self: DataProcessor, mode: str, values: list[float]) -> None:
        self.mode = mode
        self.values = values
```

## 2. Formatting Imports

Use absolute imports and avoid using of wildcard (`*`) imports.

**NOTE:** Not all imports need to be placed at the top of the module. You can import libraries inside a class if they are only used within that class, or inside conditional statements if necessary.

**NOTE:** Historically, we have accepted the use of wildcard (`*`) imports from directories within an `__init__.py` file. However, we are now phasing out this practice.

**NOTE:** Historically, we have accepted relative imports when the target is at the same level in the hierarchy inside a `__init__.py`. However, we are now phasing out this practice.

## 3. Writing Documentation

DeepTrack2 documentation should generally follow the [NumpyDoc style guide](https://numpydoc.readthedocs.io/en/latest/format.html#style-guide).

All modules, classes, methods, and functions should be documented. The documentation should include a description of the class or method, the parameters, the return value, and any exceptions that can be raised.

We sincerely appreciate any effort to improve the documentation, particularly by including examples demonstrating  how to use the classes and methods.

**NOTE:** The NumpyDoc guide specifies using phrases such as `int or list of int` rather than shorthand notations such as `int | list[int]`. We also prefer the phrases in the docstrings.

**NOTE:** In the type hints, we prefer `Literal["numpy", "torch"]` over `"numpy" | "torch"` because it is more compatible with IDEs and static type checkers.

## 1. What is `pint_definition.py`?

The `pint_definition.py` module consolidates and extends the default definitions provided by Pint's `default_en.txt` and `constants_en.txt` files. It defines physical constants, unit systems, and project-specific adjustments necessary for DeepTrack2.

## 2. Using Unit Quantities

Pint lets you include units when calculating quantities.

```python
from pint import UnitRegistry
from deeptrack.backend.pint_definition import pint_definitions
```

Initialize UnitRegistry with microscopy-related units:

```python
u = UnitRegistry(pint_definitions.split("\n"))
```

Define wavelength in nanometers:

```python
wavelength = 550 * u.nanometer  # Green light

print(f"Wavelength: {wavelength}")
```

Define speed of light in meters per second:

```python
c = 3e8 * u.meter / u.second

print(f"Speed of light: {c}")
```

Calculate and print the frequency:

```python
frequency = c / wavelength

print(f"Frequency: {frequency.to(u.terahertz)}")
```

## 2.1. Calculating the Diffraction Limit

```python
Numerical_aperture = 1.4

wavelength = 500 * u.nanometer

resolution = wavelength / (2 * Numerical_aperture)

print(f"Resolution limit: {resolution.to(u.nanometer)}")
```

## 2.2. Exposure Time and Motion Blur

```python
exposure_time = 2 * u.millisecond

sample_velocity = 10 * u.micrometer / u.second

blur_distance = sample_velocity * exposure_time

print(f"Motion blur: {blur_distance.to(u.nanometer)}")
```

## 1. What is `types.py`?

The `types` module introduces custom data type aliases for type hints throughout the DeepTrack2 codebase.
Type hints are used when declaring a function or class to improve readability and maintanability by specifying expected parameters and attribute types.

Currently there are three type aliases in DeepTrack2:

- `PropertyLike` represents a generic data type or a callable function which returns that generic datatype.
- `ImageLike` represents an image-liek structure, namely a NumPy array, a PyTorch tensor, or an `Image` obeject.
- `ArrayLike` represents an array-like structure, namley a tuple, a list, a NumPy array, a PyTorch tensor, or an `Image` obeject.
- `NumberLike` represents a numeric type, including scalars (a bool, int, float, or complex) and arrays (a NumPy array or a PyTorch tensor).

```python
from deeptrack.types import ArrayLike

def matrix_add(
    input: ArrayLike[float]
) -> ArrayLike[float]:
    """Function docsting."""
    return input + input
```

## 4. Declaring a New Type Hint

We will now declare a new type hint of our own which will represent non-integer numeric datatypes with higher precision than the standard float datatype. We will use `Union` for this, which lets us gather datatypes into a single set.

Declare the new type using Union ...

```python
import numpy as np
from typing import Union

DoubleLike = Union[np.float64, np.longdouble, np.longcomplex]
print(DoubleLike)
```

... and declare function with new type hint.

```python
def double_add(
    input: ArrayLike[DoubleLike],
    number: DoubleLike
) -> ArrayLike[DoubleLike]:
    """Function docsting."""
    return input + number
```

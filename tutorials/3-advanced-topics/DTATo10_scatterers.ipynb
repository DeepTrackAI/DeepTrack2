```python
import matplotlib.pyplot as plt
import numpy as np
```

## 1. What is a Scatterer?

Scatterers are implementations of the abstract class `Feature` that represent scattering objects (see also [DTAT301_features](DTAT301_features.ipynb)). When calculating the scattered field, they are discretized into voxels.

All physcial properties of scatterers are given in SI units, unless otherwise stated. In particular, the position of a scatterer is measured from the top left corner of the image and can either be given in meters (default) or pixels by setting the property `position_unit` to `"meter"` or `"pixel"`. The z-position of the particle can either be set as the the third element of the property `position`, or as the property `z`.

All scatterers need a reference to the optical system through which they are viewed in order to calculate scale factors. This is achieved by calling an optical device from the `optics` module (see also [DTAT323_optics](DTAT323_optics.ipynb)) with the scatterer.

## 2. Setting up the Optics Device 

Create an optical device `optics` through which to view scatterers, in this case a fluorescence microscope implemented by the `Fluorescence` class.

```python
from deeptrack.optical.optics import Fluorescence

optics = Fluorescence(
    NA=0.7,
    wavelength=680e-9,
    resolution=1e-6,
    magnification=10,
    output_region=(0, 0, 64, 64),
)
```

## 3. Creating a Point Particle

The feature `PointParticle` resolves an object occupying a single voxel, which corresponds to a point (dipolar) scatterer.

```python
from deeptrack.optical.scatterers import PointParticle

scatterer = PointParticle(
    intensity=100,  # Squared magnitude of the electric field
    position_unit="pixel",  # Units of position (default meter)
    position=(32, 32),  # Position of the particle from the upper left corner
)

imaged_scatterer = optics(scatterer)

imaged_scatterer.plot(cmap="gray")
plt.show()
```

## 4. Creating an Ellipse 

The feature `Ellipse` resolves a 2-dimensional elliptical disk, whose principal axes are in the xy-plane.

```python
from deeptrack.optical.scatterers import Ellipse

scatterer = Ellipse(
    intensity=100,  # Squared magnitude of the electric field.
    position_unit="pixel",  # Units of position (default meter).
    position=(32, 32),  # Position of the particle.
    radius=(1e-6, 0.5e-6),  # Dimensions of the principal axes of the ellipse.
    rotation=np.pi / 4,  # Amount the ellipse is rotated by in radians.
    upsample=4,  # Amount the resolution is upsampled for accuracy.
)

imaged_scatterer = optics(scatterer)

imaged_scatterer.plot(cmap="gray")
plt.show()
```

## 5. Creating a Sphere

The feature `Sphere` resolves a 3-dimensional sphere.

```python
from deeptrack.optical.scatterers import Sphere

scatterer = Sphere(
    intensity=100,  # Squared magnitude of the electric field.
    position_unit="pixel",  # Units of position (default meter).
    position=(32, 32),  # Position of the particle.
    radius=1e-6,  # Radius of the sphere.
    upsample=4,  # Amount the resolution is upsampled for accuracy.
)

imaged_scatterer = optics(scatterer)

imaged_scatterer.plot(cmap="gray")
plt.show()
```

## 6. Creating an Ellipsoid

The feature `Ellipsoid` resolves a 3-dimensional ellipsoid.

```python
from deeptrack.optical.scatterers import Ellipsoid

scatterer = Ellipsoid(
    intensity=100,  # Squared magnitude of the electric field.
    position_unit="pixel",  # Units of position (default meter).
    position=(32, 32),  # Position of the particle.
    radius=(1e-6, 0.5e-6, 0.25e-6),  # Principal axes dimensions.
    rotation=(np.pi / 4, 0, 0),  # Rotation in radians.
    upsample=4,  # Resolution upsampling for accuracy.
)

imaged_scatterer = optics(scatterer)

imaged_scatterer.plot(cmap="gray")
plt.show()
```

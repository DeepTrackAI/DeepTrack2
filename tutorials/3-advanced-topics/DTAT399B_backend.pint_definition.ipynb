This advanced tutorial introduces the pint_definition module.

## 1. What is `pint_definition`?

The `pint_definition` module consolidates and extends the default definitions provided by Pint's 
`default_en.txt` and `constants_en.txt` files. It defines physical constants, 
unit systems, and project-specific adjustments necessary for DeepTrack2.

## 2. Unit Quantities
Pint lets us include units when calculating quantities.


```python
from pint import UnitRegistry
from deeptrack.backend.pint_definition import pint_definitions

# Initialize UnitRegistry with microscopy-related units
units = UnitRegistry(pint_definitions.split("\n"))

# Define wavelength in nanometers
wavelength = 550 * units.nanometer  # Green light

c = 3e8 * units.meter / units.second

frequency = c / wavelength

print(f"Wavelength: {wavelength}")
print(f"Frequency: {frequency.to(units.terahertz)}") 
```

## 3. Diffraction Limit

```python
Numerical_aperture = 1.4 
wavelength = 500 * units.nanometer
resolution = wavelength / (2 * Numerical_aperture)

print(f"Resolution limit: {resolution.to(units.nanometer)}")
```

## 4. Exposure Time and Motion Blur

```python
exposure_time = 2 * units.millisecond  

sample_velocity = 10 * units.micrometer / units.second 

blur_distance = sample_velocity * exposure_time

print(f"Motion blur: {blur_distance.to(units.nanometer)}")
```

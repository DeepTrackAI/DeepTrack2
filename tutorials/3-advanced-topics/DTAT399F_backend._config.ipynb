## 1. What is `_config.py`?

The `_config` module is DeepTrack2's central component for configuration and backend management.

It provides a unified interface to control key computational aspects of DeepTrack2 pipelines, ensuring flexibility and consistency across different environments and user preferences.

The key roles of `_config.py`are:

- **Backend Management:**
    It allows users to select and seamlessly switch between different computational backends—primarily NumPy and PyTorch—so that DeepTrack2 code can remain backend-agnostic and portable.

- **Device Selection:**
    It enables explicit selection of computation devices (e.g., CPU, GPU, or even a specific PyTorch device object), allowing users to use available hardware resources efficiently.

- **Global Configuration Object:**
    It exposes a singleton `config` object, which maintains the current backend and device settings. This ensures that all parts of DeepTrack2 share a consistent configuration state.

- **Backend Proxy (`xp`):**
    It provides the `xp` proxy object, which forwards array creation and manipulation functions to the currently selected backend. This means you can write code that is agnostic to the actual backend in use, without manual imports of NumPy or PyTorch.

- **Dependency Availability Checks:**
    Tracks whether optional dependencies like PyTorch and OpenCV are available, enabling code to gracefully degrade or branch as needed.

## 2. Importing the `config` Object and Checking the Current Backend

You can import the global `config` object with the following command:

```python
from deeptrack.backend._config import config
```

You can then check the default backend ...

```python
print(config.get_backend())
```

... and the default device.

```python
print(config.get_device())
```

### 2.1. Switching the Computational Backend

**NOTE:** You should ensure the backend is correctly installed.

You can switch the computational backend to use PyTorch ...

```python
config.set_backend("torch")
# config.set_backend_torch()  # Alternative form

print(config.get_backend())
```

... and back to use NumPy.

```python
config.set_backend("numpy")
# config.set_backend_numpy()  # Alternative form

print(config.get_backend())
```

You can also temporarily switch backends within a context manager.

```python
print(config.get_backend())
```

```python
with config.with_backend("numpy"):
    print(config.get_backend())
```

```python
print(config.get_backend())
```

### 2.2. Switching the Computational Device

**NOTE:** You should ensure the device exists and is correctly installed.

You can switch to use the GPU with one of the following codes.

```python
config.set_device("cuda")

print(config.get_device())
```

```python
config.set_device("mps")

print(config.get_device())
```

```python
config.set_device("cuda:0")

print(config.get_device())
```

```python
import torch

config.set_device(torch.device("cuda:0"))

print(config.get_device())
```

## 3. Importing the `xp` Proxy

You can import the `xp` proxy for backend-agnostic code with the following command:

```python
from deeptrack.backend._config import xp
```

Use the `xp` proxy to create a NumPy array ...

```python
config.set_backend_numpy()

array = xp.arange(5)

print(type(array))
```

... or a PyTorch tensor.

```python
config.set_backend_torch()

tensor = xp.arange(3)

print(type(tensor))
```
